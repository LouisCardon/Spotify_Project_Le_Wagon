```python
songs['percentagelikeinviews'] = songs['Likes'] / songs['Views']*100
```

```python
consolidated_df = pd.merge(left =Top_100, right = songs,how='left', left_on= "title", right_on = "Track" )
```

```python
consolidated_df.head(3)
```

```python
consolidated_df.shape
```

```python
consolidated_df = consolidated_df.dropna(subset=['percentagelikeinviews'])
```

```python
consolidated_df["top genre"].unique()
```

```python
def assign_category(genre):
    if 'pop' in genre.lower(): 
        return 'Pop'
    elif 'rock' in genre.lower(): 
        return 'Rock'
    else:
        return genre
```

```python
consolidated_df["Genre_Category"] = consolidated_df["top genre"].apply(assign_category)
```

```python
px.scatter(consolidated_df, x="Views", y="Likes", color="Genre_Category", size='percentagelikeinviews', hover_data=['Artist'])
```
